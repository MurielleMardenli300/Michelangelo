```python
import numpy
import os

def analyze_obj(file_path):
    # Initialize metric counters
    stats = {
        "vertices": 0,
        "texture_coords": 0,
        "normals": 0,
        "faces": 0,
        "triangles": 0,
        "quads": 0,
        "polygons": 0
    }
    
    # Initialize bounding box limits
    min_x = min_y = min_z = float('inf')
    max_x = max_y = max_z = float('-inf')

    try:
        with open(file_path, 'r') as file:
            for line in file:
                # Clean line and split by whitespace
                tokens = line.strip().split()
                if not tokens:
                    continue
                
                prefix = tokens[0]

                # Process vertices
                if prefix == 'v':
                    stats["vertices"] += 1
                    x, y, z = float(tokens[1]), float(tokens[2]), float(tokens[3])
                    min_x, max_x = min(min_x, x), max(max_x, x)
                    min_y, max_y = min(min_y, y), max(max_y, y)
                    min_z, max_z = min(min_z, z), max(max_z, z)
                
                # Process texture coordinates
                elif prefix == 'vt':
                    stats["texture_coords"] += 1
                
                # Process vertex normals
                elif prefix == 'vn':
                    stats["normals"] += 1
                
                # Process faces
                elif prefix == 'f':
                    stats["faces"] += 1
                    num_vertices_in_face = len(tokens) - 1
                    if num_vertices_in_face == 3:
                        stats["triangles"] += 1
                    elif num_vertices_in_face == 4:
                        stats["quads"] += 1
                    elif num_vertices_in_face > 4:
                        stats["polygons"] += 1

        # Print results summary
        print(f"--- Analysis Summary for: {file_path} ---")
        print(f"Vertices:           {stats['vertices']}")
        print(f"Texture Coords:     {stats['texture_coords']}")
        print(f"Vertex Normals:     {stats['normals']}")
        print(f"Total Faces:        {stats['faces']} (Triangles: {stats['triangles']}, Quads: {stats['quads']}, Polygons: {stats['polygons']})")
        
        if stats["vertices"] > 0:
            print("\n--- Bounding Box ---")
            print(f"X range: [{min_x:.4f}, {max_x:.4f}] -> Size: {max_x - min_x:.4f}")
            print(f"Y range: [{min_y:.4f}, {max_y:.4f}] -> Size: {max_y - min_y:.4f}")
            print(f"Z range: [{min_z:.4f}, {max_z:.4f}] -> Size: {max_z - min_z:.4f}")

    except FileNotFoundError:
        print(f"Error: The file '{file_path}' was not found.")
    except Exception as e:
        print(f"An error occurred during analysis: {e}")

# Example usage:
# analyze_obj("your_model.obj")
```

```python
analyze_obj("reconstruction/reconstruction.obj")
```
